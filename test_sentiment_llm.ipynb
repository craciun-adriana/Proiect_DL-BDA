```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset(
    "fancyzhx/amazon_polarity",
    split="train[:10]"
)

df = pd.DataFrame(dataset)
df = df.rename(columns={
    "content": "text",
    "label": "original_label"
})

df["original_label"] = df["original_label"].map({
    0: "negative",
    1: "positive"
})

df.insert(0, "review_id", range(1, len(df) + 1))

print("Număr de recenzii:", len(df))
display(df[["review_id", "text", "original_label"]])
```

```python

from transformers import pipeline

sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

results = sentiment_model(
    df["text"].fillna("").astype(str).tolist(),
    truncation=True,
    max_length=512
)

df["model_sentiment"] = [
    "positive" if result["label"] == "POSITIVE" else "negative"
    for result in results
]

df["confidence"] = [
    round(result["score"], 4) for result in results
]

display(
    df[["review_id", "text", "original_label",
        "model_sentiment", "confidence"]]
)
```

```python

from datasets import load_dataset
import pandas as pd

# Încărcăm datele din Amazon Polarity
dataset = load_dataset(
    "fancyzhx/amazon_polarity",
    split="train"
)

# Selectăm aleatoriu 5.300 de recenzii
sample = dataset.shuffle(seed=42).select(range(5300))

df_all = pd.DataFrame(sample)
df_all = df_all.rename(
    columns={
        "content": "text",
        "label": "original_label"
    }
)

# Păstrăm etichetele originale separat
df_all["original_label"] = df_all["original_label"].map(
    {0: "negative", 1: "positive"}
)

# ID unic pentru fiecare recenzie
df_all.insert(0, "review_id", range(1, len(df_all) + 1))

# Eliminăm eventualele texte lipsă
df_all = df_all.dropna(subset=["text"]).reset_index(drop=True)

print("Număr de recenzii:", len(df_all))
display(df_all.head())
```

```python

# Primele 300 de recenzii vor fi etichetate manual
manual_df = df_all.iloc[:300][
    ["review_id", "text", "original_label"]
].copy()

manual_df["manual_label"] = ""

# Restul de 5.000 vor fi etichetate ulterior prin LLM
llm_df = df_all.iloc[300:][
    ["review_id", "text", "original_label"]
].copy()

print("Set pentru etichetare manuală:", len(manual_df))
print("Set pentru etichetare LLM:", len(llm_df))

print("\nExemplu din setul manual:")
display(manual_df.head())

print("\nExemplu din setul LLM:")
display(llm_df.head())
```

```python

# Salvăm setul complet
df_all.to_parquet(
    "amazon_reviews_sample.parquet",
    index=False
)

# Salvăm setul pentru LLM
llm_df.to_parquet(
    "amazon_reviews_llm.parquet",
    index=False
)

# CSV pentru etichetarea manuală
manual_df.to_csv(
    "amazon_reviews_manual.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Fișierele au fost salvate.")
```
